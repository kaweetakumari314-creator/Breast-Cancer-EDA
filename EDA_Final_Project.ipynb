**8. Multivariate Analysis**

```python
EP_positive = df[(df['Estrogen Status']=='Positive') & (df['Progesterone Status']=='Positive')]
```

```python
print(EP_positive['Age'].value_counts())
```

```python
sns.histplot(data=EP_positive, x='Age', kde=True, color='purple')
plt.title('Age Distribution of Estrogen and Progesterone Positive Patients')
plt.show()
```

```python

# Group by cancer stage and estrogen status
grouped_data = df.groupby(
    ["6th Stage", "Estrogen Status"]
)["Survival Months"].mean().reset_index()

print(grouped_data)


# Pivot table
pivot_table = grouped_data.pivot(
    index="6th Stage",
    columns="Estrogen Status",
    values="Survival Months"
)

print(pivot_table)


# Heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    pivot_table,
    annot=True,
    cmap="YlGnBu",
    fmt=".1f"
)

plt.title("Average Survival by Cancer Stage and Estrogen Status")
plt.xlabel("Estrogen Status")
plt.ylabel("Cancer Stage")

plt.show()
```

```python
numerical_cols = ['Age', 'Tumor Size', 'Regional Node Examined', 'Reginol Node Positive', 'Survival Months']

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers_count = len(df[(df[col] < lower_bound) | (df[col] > upper_bound)])
    print(f"Column '{col}' has {outliers_count} outliers.")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Plot Age, Tumor Size, and Survival Months
sns.boxplot(ax=axes[0], y=df['Age'], color='skyblue').set_title('Age Outliers')
sns.boxplot(ax=axes[1], y=df['Tumor Size'], color='salmon').set_title('Tumor Size Outliers')
sns.boxplot(ax=axes[2], y=df['Survival Months'], color='lightgreen').set_title('Survival Months Outliers')

plt.tight_layout()
plt.show()
```

```python
# Calculate bounds for Survival Months only
Q1 = df['Survival Months'].quantile(0.25)
Q3 = df['Survival Months'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Keep only rows where Survival Months is within bounds
df_clean_survival = df[(df['Survival Months'] >= lower_bound) & (df['Survival Months'] <= upper_bound)]

print(f"Rows remaining after cleaning Survival Months: {len(df_clean_survival)}")
```

```python
columns_to_clean = ['Tumor Size', 'Regional Node Examined', 'Reginol Node Positive', 'Survival Months']


df_clean = df.copy()


for col in columns_to_clean:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_clean = df_clean[(df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)]


print(f"Original dataset rows: {len(df)}")
print(f"Cleaned dataset rows: {len(df_clean)}")
print(f"Total rows removed: {len(df) - len(df_clean)}")
```

```python
plt.figure(figsize=(8, 6))


numeric_df = df_clean[['Age', 'Tumor Size', 'Regional Node Examined', 'Reginol Node Positive', 'Survival Months']]


corr_matrix = numeric_df.corr()


sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix of Patient Metrics')
plt.show()
```
